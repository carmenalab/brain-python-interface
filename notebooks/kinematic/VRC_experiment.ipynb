# Default settings

```python
# Generate default settings

from riglib.experiment.scripting import run_experiment

project = "VRC"
task = "manual control"

features = ["autostart",
     "optitrack",
     "penalty_audio",
     "rand_delay",
     "reward_audio",
     "saveHDF",
     "syncHDF",
     "windowVR"]

parameters = {
    "scale": 1,
    "offset": [-20, -95, -2],
    "reward_sound": "incorrect.wav",
    "penalty_sounds": "buzzer.wav",
    "rand_start": [0.25, 0.75],
    "rand_delay": [0,0],
    "prob_catch_trials": 0,
    "grid_size": 130,
    "camera_position": [0, 0, -40],
    "fixed_camera_position": False,
    "fixed_camera_orientation": False,
    "fullscreen": True,
    "session_length": 0,
    "velocity_control": False,
    "random_rewards": False,
    "rotation": "yzx",
    "exp_rotation": "none",
    "exp_gain": 1.0,
    "limit2d": False,
    "target_radius": 1,
    "plant_type": "cursor",
    "plant_visible": True,
    "reward_time": 0.5,
    "hold_time": 0.2,
    "hold_penalty_time": 1,
    "delay_penalty_time": 1,
    "timeout_time": 10,
    "timeout_penalty_time": 1,
    "max_attempts": 3,
    "num_targets_per_attempt": 2,
    "screen_dist": 80
    }

generator = "centerout_2d"

parameters['grid_position'] = [0, 0, 130-eye_height]
parameters['camera_offset'] = [0, -eye_height, 50]

if group==1: # 2D feedback
    parameters['cursor_bounds'] = [-14,14,0,0,-14,14]
elif group==2: # 3D feedback
    parameters['cursor_bounds'] = [-14,14,-14,14,-14,14]
    features.append("cylinder_center_out")

settings = {"features": features, "parameters": parameters}

import pickle
with open('VRC_defaults.pkl', 'wb') as f:
    pickle.dump(settings, f)

from IPython.display import HTML
def play_sound(file):
    return HTML(f"""
    <audio autoplay>
      <source src="{file}" type="audio/mpeg">
    </audio>
    """)
```

# Baseline Session
* Let participants know that the cursor will now be controlled with their hand motions.
* Don't give further details on the control mapping. If they ask, just say: "That's not something that I can tell you."

```python
# Block 1 (a1)

with open('VRC_defaults.pkl', 'rb') as f:
    settings = pickle.load(f)
    
parameters = settings['parameters']
features = settings['features']
    
session = "a1"
nblocks = 8

if group==1: # 2D feedback
    task_desc = "Baseline 2D feedback"
elif group==2: # 3D feedback
    task_desc = "Baseline 3D feedback"
    
sequence = generator + ":[nblocks=" + str(nblocks) + ", ntargets=8, distance=10, origin=0 0 0]"
run_experiment(subject,experimenter,project,session,task,features,sequence,task_desc,save,**parameters)

print("First baseline block (a1) completed.")
play_sound("Bell.wav")
```

```python
# Block 2 (a2)

session = "a2"

run_experiment(subject,experimenter,project,session,task,features,sequence,task_desc,save,**parameters)

print("Second baseline block (a2) completed.")
play_sound("Bell.wav")
```

```python
# Block 3 (a3)

session = "a3"

run_experiment(subject,experimenter,project,session,task,features,sequence,task_desc,save,**parameters)

print("Third baseline block (a3) completed.")
play_sound("Bell.wav")
```
