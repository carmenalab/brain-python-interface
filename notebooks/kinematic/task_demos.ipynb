```python
experimenter = "Victoria"

# Seated eye height of subject
eye_height = 130 # measure in cm



#### ---- Do not edit below here --- ####
subject = "test"
save = False # don't save data
if not experimenter[0].isupper():
    print("Warning: Capitalize your name for consistency. Check spelling.")
```

# Default settings

```python
# Generate default settings

from riglib.experiment.scripting import run_experiment
from datetime import datetime
import os

os.environ['BMI3D_PORT'] = "8000"

project = "demos"

features = ["autostart",
     "optitrack",
     "penalty_audio",
     "rand_delay",
     "reward_audio",
     "saveHDF",
     "syncHDF"]

parameters = {
    "scale": 1,
    "offset": [-20, -95, -2],
    "reward_sound": "incorrect.wav",
    "penalty_sounds": "buzzer.wav",
    "rand_start": [0.25, 0.75], # [2,2]?
    "rand_delay": [0,0],
    "prob_catch_trials": 0,
    "fullscreen": True,
    "session_length": 0,
    "velocity_control": False,
    "random_rewards": False,
    "rotation": "yzx",
    "exp_rotation": "none",
    "pertubation_rotation": 0,
    "perturbation_rotation_x": 0,
    "perturbation_rotation_z": 0,
    "exp_gain": 1.0,
    "target_radius": 2.5,
    "plant_type": "cursor",
    "plant_visible": True,
    "reward_time": 0.5, # 0.2?
    "hold_time": 0.2,
    "hold_penalty_time": 1,
    "delay_penalty_time": 1,
    "timeout_time": 10,
    "timeout_penalty_time": 1,
    "max_attempts": 2,
    "num_targets_per_attempt": 2 # 1?
    }

settings = {"features": features, "parameters": parameters}

import pickle
with open('demo_defaults.pkl', 'wb') as f:
    pickle.dump(settings, f)

from IPython.display import HTML
def play_sound(file):
    return HTML(f"""
    <audio autoplay>
      <source src="{file}" type="audio/mpeg">
    </audio>
    """)
```

# 3D Center-Out

## Upright-45
* Upright display, 45-degree in-plane perturbation

```python
task = "manual control"

with open('demo_defaults.pkl', 'rb') as f:
    settings = pickle.load(f)
    
parameters = settings['parameters']
features = settings['features']

features.append("windowVR")
    
session = "demo"
nblocks = 4

generator = "centerout_2d"

parameters["show_grid"] = True
parameters["grid_size"] = 130
parameters["fixed_camera_position"]: False
parameters["fixed_camera_orientation"]: False
parameters["limit2d"] = False

parameters['cursor_bounds'] = [-14,14,0,0,-14,14]
parameters['grid_position'] = [0, 0, 130-eye_height]
parameters['camera_offset'] = [0, -eye_height, 50]

parameters['exp_rotation'] = "none"
parameters['pertubation_rotation'] = 45

task_desc = "3D upright centerout, in-plane 45 perturbation"
    
sequence = generator + ":[nblocks=" + str(nblocks) + ", ntargets=8, distance=10, origin=0 0 0]"
run_experiment(subject,experimenter,project,session,task,features,sequence,task_desc,save,**parameters)

now = datetime.now()

print(f'Demo completed at {now.strftime("%I:%M %p, %m-%d-%Y")}.')
play_sound("Bell.wav")
```
